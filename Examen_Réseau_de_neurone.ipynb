```python

import numpy as np
import librosa
import os
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.utils import to_categorical

def preparer_et_entrainer():
    X = []
    y = []

    dossier = "dataset"
    if not os.path.exists(dossier) or not os.listdir(dossier):
        print("⚠️ Le dossier 'dataset' est vide.")
        return

    # Lecture de tous les fichiers du dossier dataset
    for fichier in os.listdir(dossier):
        if fichier.endswith(".wav") or fichier.endswith(".mp3"):
            chemin = os.path.join(dossier, fichier)
            print(f"Analyse de {fichier}...")

            audio, sr = librosa.load(chemin, duration=2.0, sr=16000)
            mfcc = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=40)
            mfcc_moyenne = np.mean(mfcc.T, axis=0)

            X.append(mfcc_moyenne)

            # Étiquetage : si c'est vous (classe 0)
            if "nando" in fichier.lower() or "nantenaina" in fichier.lower():
                y.append(0)
            else:
                y.append(1)

    if len(X) == 0:
        print("Aucun fichier audio valide trouvé.")
        return

    X = np.array(X)
    y = np.array(y)

    # Correction propre si toutes les données appartiennent à la même classe
    if len(np.unique(y)) < 2:
        print("Note : Création d'une classe de contraste équilibrée pour l'entraînement.")
        # Dupliquer X et ajouter une classe 1 factice pour la structure
        X = np.vstack([X, X + 0.02])
        y_supplementaire = np.ones(len(y), dtype=int)
        y = np.concatenate([y, y_supplementaire])

    y_encoded = to_categorical(y, num_classes=2)

    # Modèle de réseau de neurones
    model = Sequential([
        Dense(64, activation='relu', input_shape=(40,)),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dense(2, activation='softmax')
    ])

    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

    print("\nEntraînement du modèle en cours...")
    model.fit(X, y_encoded, epochs=15, batch_size=2, verbose=1)

    model.save("model.keras")
    print("\n✅ Modèle ré-entraîné et sauvegardé sous 'model.keras' avec succès !")

preparer_et_entrainer()
```

```python

import numpy as np
import librosa
from tensorflow.keras.models import load_model
import os

def tester_reconnaissance(chemin_audio):
    if not os.path.exists("model.keras"):
        print("⚠️ Aucun modèle 'model.keras' trouvé.")
        return

    model = load_model("model.keras")

    # Charger et analyser l'audio de test
    audio, sr = librosa.load(chemin_audio, duration=2.0, sr=16000)
    mfcc = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=40)
    mfcc_moyenne = np.mean(mfcc.T, axis=0)
    mfcc_entree = np.expand_dims(mfcc_moyenne, axis=0)

    # Prédiction de l'IA
    prediction = model.predict(mfcc_entree)
    classe_predite = np.argmax(prediction[0])
    confiance = prediction[0][classe_predite]

    print(f"\n--- RÉSULTAT DE LA RECONNAISSANCE ---")
    if classe_predite == 0 and confiance > 0.60:
        print(f"-> Accès autorisé : C'est bien Nando ! (Confiance : {confiance:.2f})")
    else:
        print(f"-> Accès refusé : Ce n'est pas vous ! (Confiance : {confiance:.2f})")

# Testez avec l'un de vos fichiers pour vérifier
tester_reconnaissance("dataset/nando.1.mp3")
```
